```python
import numpy as np
np.random.seed(42)
marks = np.array([55, 72, 48, 90, 66, 81])
print('Marks:', marks)
print('Mean:', marks.mean())
print('Highest:', marks.max())
print('Passed:', marks[marks >= 50])

#class work
#Minimum & Standard Deviation
print('Minimum:', marks.min())
print('Standard Deviation:', marks.std())

#adding the bonus mark
bonus_mark = marks+5
print('Marks:', marks)
print('Bonus Marks:', bonus_mark)

#counting passed student
passed_students = np.sum(marks >= 50)
print('Number of Passed Students:', passed_students)

#Create a 2 × 3 matrix from the marks.
matrix = np.array([[55, 72, 48], [90, 66, 81]])
print('Matrix:')
print(matrix)
```

```python
import pandas as pd
data = {
'student': ['Asha', 'Babu', 'Chayan', 'Nibir', 'Emon', 'Farah'],
'study_hours': [2, 5, 1, 7, 3, 6],
'sleep_hours': [8, 6, 9, 5, 7, 6],
'attendance': [70, 88, 55, 95, 75, 91],
'snacks': [3, 2, 5, 1, 4, 2],
'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}
df = pd.DataFrame(data)
display(df)
print(df.info())
print(df.select_dtypes(include='number').describe())

#How many samples and columns are present?
print('Number of Samples:', df.shape[0])
print('Number of Columns:', df.shape[1])

#Filter students with attendance of at least 80.
filtered_df = df[df['attendance'] >= 80]
display(filtered_df)

#Calculate average study hours for students who passed.
passed_df = df[df['passed'] == 'Yes']
average_study_hours = passed_df['study_hours'].mean()
print('Average Study Hours for Passed Students:', average_study_hours)

#Sort the DataFrame by attendance in descending order.
sorted_df = df.sort_values(by='attendance', ascending=False)
display(sorted_df)

```

```python
#Does this tiny dataset prove that snack purchases cause failure?
print("""
Answer: No, this dataset does NOT prove that snacks cause failure.
It only shows a possible correlation, not causation. With only 6 samples,
students who snack more (Chayan, Emon) also happen to have the lowest
study hours and attendance — so low study time/attendance is a more
likely cause of failure, with snacking being an unrelated side habit,
not the actual cause. Proving causation would require a controlled
experiment isolating snacks as the only variable, plus a much larger sample size.
""")
```
